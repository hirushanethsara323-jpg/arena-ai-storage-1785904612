## Step 3: Free Unlimited Grok via Puter.js (No API Keys!) - The Magic

Instead of `import openai` with API key, we use **Puter.js** in Colab via JavaScript!

Puter.js works in HTML: `<script src="https://js.puter.com/v2/"></script>` then `puter.ai.chat()` with `model: 'x-ai/grok-4.5'`

In Colab, we can display HTML with Puter.js and chat with Grok for free!


```python
from IPython.display import HTML, display

html = """
<script src="https://js.puter.com/v2/"></script>
<div id=\"grok-output\" style=\"background:#0A0A0F;color:#00FFD1;padding:16px;border-radius:12px;font-family:monospace;white-space:pre-wrap\">Loading Grok 4.5 via Puter.js (Free Unlimited)...</div>
<script>
(async () => {
  try {
    const response = await puter.ai.chat(
      "Explain Zero OS v9.0 architecture: 84K kernel, real bootable 74K 32-bit ELF, QEMU tested, FAT32 real disk.img 10MB with boot sig 0x55AA, USB-HID queue 128 usbtype inject, VESA 1024x768x32 LFB at 0xFD000000, PMM bitmap bug fix 0x10000->0x70000, PCI outl/inl fix triple fault, ATA inw/outw fix. Be witty.",
      {model: 'x-ai/grok-4.5', stream: true}
    );
    let full='';
    const el = document.getElementById('grok-output');
    for await (const part of response) {
      if(part.text){ full+=part.text; el.textContent=full; }
    }
  } catch(e) {
    document.getElementById('grok-output').textContent = 'Error: ' + e.message + '\\nMake sure signed into Puter.com (free) - https://docs.puter.com/getting-started/';
  }
})();
</script>
"""
display(HTML(html))
```
