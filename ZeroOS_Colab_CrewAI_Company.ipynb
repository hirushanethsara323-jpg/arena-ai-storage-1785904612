## Step 4: Define CrewAI Company - Agents & Tasks for Zero OS

```python
from crewai import Agent, Task, Crew, Process
from crewai_tools import FileReadTool, FileWriteTool, DirectoryReadTool
import os

os.chdir("/content/ZeroOS_Repo/ZeroOS")

# Tools
read_tool = FileReadTool()
write_tool = FileWriteTool()
dir_tool = DirectoryReadTool(directory="/content/ZeroOS_Repo/ZeroOS")

# === AGENTS - Company Structure ===

ceo = Agent(
    role="CEO of Zero OS",
    goal="Define vision: Zero Bloat, Zero Tracking, Zero Limits. Roadmap v6.0 to v10.0",
    backstory="You are CEO of Zero OS startup. You started from zero, built real bootable OS. You decide next features.",
    verbose=True,
    allow_delegation=False
)

cto = Agent(
    role="CTO / OS Architect",
    goal="Design GDT, IDT, Paging, PMM, Heap, overall architecture for real bootable OS",
    backstory="You are CTO, expert in x86_64, multiboot, VESA, paging with higher half. You fixed PMM bitmap overwriting multiboot info bug at 0x10000 -> 0x70000.",
    verbose=True,
    tools=[read_tool, dir_tool],
    allow_delegation=False
)

kernel_dev = Agent(
    role="Kernel Developer",
    goal="Develop boot.S, kernel.c, serial driver, PMM, Heap, PIT, Tasking",
    backstory="You write freestanding C, no stdlib, boot_real.S with VIDEO flag 1024x768x32, serial 0x3F8 for QEMU debug. You built 74K 32-bit real bootable kernel.",
    verbose=True,
    tools=[read_tool, write_tool],
    allow_delegation=False
)

driver_dev = Agent(
    role="Driver Developer",
    goal="Develop all drivers: PCI outl/inl fix triple fault, USB UHCI QH/TD, USB-HID real queue, ATA inw/outw fix, AC97 BAR0, NET NE2000/E1000, SMP APIC, Speaker, Mouse, FB VESA",
    backstory="You fixed PCI triple fault by changing 4x outb to outl/inl, ATA inb to inw, PMM reserve kernel+heap to avoid overwrite. You made FAT32 real disk read work in QEMU.",
    verbose=True,
    tools=[read_tool, write_tool],
    allow_delegation=False
)

fs_engineer = Agent(
    role="Filesystem Engineer",
    goal="Develop ZeroFS v1 32 files, ZeroFS2 v2 128 inodes journal, FAT32 real mount with ATA, disk.img 10MB with boot sig 0x55AA",
    backstory="You created disk.img 10MB FAT32 via Python, boot sector, 2 FATs 128 sectors, root cluster 2 LBA 288 with README 95 bytes, ZERO 68 bytes. QEMU -hda disk.img real read proven: FAT32 Root README.TXT ZERO.TXT",
    verbose=True,
    tools=[read_tool, write_tool],
    allow_delegation=False
)

gui_engineer = Agent(
    role="GUI Engineer - Zero Ring Unique Design",
    goal="Develop FB VESA parse mb_info flags bit12, font 8x8 bitmap, compositor Zero Ring with orbit apps, window drag, close button, mouse cursor 8x8",
    backstory="You designed unique Zero Ring: no taskbar, center 0 click for orbit, 6 apps orbit positions, drag windows, drop to center to close, colors #0A0A0F bg #00FFD1 cyan #8B5CF6 purple.",
    verbose=True,
    tools=[read_tool, write_tool],
    allow_delegation=False
)

userland_engineer = Agent(
    role="Userland Engineer",
    goal="Develop Tasking 16 tasks preemptive PIT IRQ0, ELF loader 32-bit, Syscalls int 0x80, App Store 6 apps, Shell 40+ commands, Calculator app calc_eval 2+2, AI local",
    backstory="You built shell with 40+ commands: ls cat touch rm write ls2 touch2 fatls disk gui ps ticks spawn exec apps launch store paging smp beep play pci usb usbtype net ping ai calc. Calc app does 2+2 etc.",
    verbose=True,
    tools=[read_tool, write_tool],
    allow_delegation=False
)

qa_tester = Agent(
    role="QA Tester - QEMU Real Boot",
    goal="Test QEMU real bootable: qemu-system-i386 -kernel zero-kernel-32.elf -hda disk.img -net nic,model=e1000 -usb -device usb-ehci -device usb-kbd -vga std -nographic, check serial log for VESA OK and shell prompt",
    backstory="You installed QEMU 10.0.11 via apt, built 74K 32-bit ELF with gcc-multilib -m32, ISO 10MB via grub-mkrescue, proven boot log: Multiboot magic 0x2BADB002, GDT OK, Paging PD @0x00300000, PCI Found 6-8 devices, FAT32 Mounted real BPS 512, shell zero@zero-os:~$",
    verbose=True,
    tools=[read_tool],
    allow_delegation=False
)

devops = Agent(
    role="DevOps - GitHub Storage",
    goal="Manage GitHub as storage: arena-ai-storage-1785904612 repo, handle .netrc token, 128MB sandbox limit vs GitHub unlimited GBs, push ISO and disk.img, storage no issue 2.3MB used",
    backstory="You connected GitHub as storage via .netrc, token ghp_ZVO..., repo hirushanethsara323-jpg/arena-ai-storage-1785904612, push 21MB with ISO+disk, sandbox 13-32MB /128MB 25% free.",
    verbose=True,
    tools=[read_tool, dir_tool],
    allow_delegation=False
)

print("✅ 9 Agents created - Zero OS Company ready!")
```

## Step 5: Define Tasks for Next Version (v10.0)

```python
task1 = Task(
    description="CEO defines v10.0 roadmap: After v9.0 84K ALL EVERYTHING with calc app, what next? Options: v10 Daily Driver Installer with Ext4, USB mass storage, TCP/IP real ping 8.8.8.8, SMP AP boot multicore, AC97 PCM real wav. Decide and document in ROADMAP.md",
    expected_output="ROADMAP.md with v10.0 features list",
    agent=ceo
)

task2 = Task(
    description="CTO designs paging higher half 0xC0000000 + enable real, TSS, real context switch with fxsave, higher half linker script, and VESA boot to GUI by default. Check current paging.c PMM reserve bug fix and fb.c VESA parse.",
    expected_output="Design doc for paging higher half and context switch",
    agent=cto
)

task3 = Task(
    description="Kernel Developer: Build real 32-bit kernel with -m32, fix any -Werror, ensure zero-kernel-32.elf 74K+ builds, test QEMU -kernel boot to shell, serial log shows all OK",
    expected_output="zero-kernel-32.elf built and QEMU boot log",
    agent=kernel_dev
)

task4 = Task(
    description="Driver Developer: Perfect all drivers - USB HID real interrupt TD completion not just queue inject, ATA PIO with inw/outw already fixed, PCI outl/inl fixed triple fault, AC97 PCM buffer real, NET e1000 TX/RX descriptors, SMP AP trampoline 0x8000 IPI via LAPIC 0xFEE00000",
    expected_output="Driver code improvements for USB, ATA, AC97, NET, SMP",
    agent=driver_dev
)

task5 = Task(
    description="Filesystem Engineer: Improve FAT32 write/create/delete real cluster chain + FAT table update via ATA write, ZeroFS3 with journaling, test with disk.img 10MB: create file via touch then sync to ATA sector",
    expected_output="FAT32 write support and ZeroFS3 design",
    agent=fs_engineer
)

task6 = Task(
    description="GUI Engineer: Make GUI boot real with mouse cursor 8x8 white/black, window drag with offset, close button, orbit apps launch on ring click, font 8x8 draw. Currently fb.c VESA 1024x768x32 LFB at 0xFD000000 pitch 4096 proven, but need to fix PMM bitmap overwrite bug (0x10000 -> 0x70000) which you already did.",
    expected_output="GUI improvements with mouse and window drag working in QEMU",
    agent=gui_engineer
)

task7 = Task(
    description="Userland Engineer: Add more apps: Calculator already does calc 2+2, add Editor with save to FAT32, File Manager GUI listing ZeroFS and FAT32 real files from disk.img, Browser with HTTP via NET, improve shell with 40+ commands including calc, usbtype, disk, ai, ping",
    expected_output="New apps: calc, editor, files GUI improvements",
    agent=userland_engineer
)

task8 = Task(
    description="QA Tester: Run QEMU tests: 1) qemu-system-i386 -kernel zero-kernel-32.elf -nographic check serial log shows VESA OK, Shell prompt, FAT32 Mounted real, 2) with -hda disk.img check FAT32 Root README.TXT ZERO.TXT real disk read, 3) with -net nic,model=e1000 check PCI Found NIC, 4) ISO boot qemu-system-i386 -cdrom zero-os.iso -boot d -vga std -display none -serial stdio check VESA 1024x768x32 LFB at 0xFD000000",
    expected_output="QEMU test logs for all configs",
    agent=qa_tester
)

task9 = Task(
    description="DevOps: Push to GitHub repo hirushanethsara323-jpg/arena-ai-storage-1785904612, manage .netrc token, handle storage: sandbox 13-32MB /128MB 25% free, GitHub 21MB with ISO+disk, no issue, ensure ISO and disk.img committed, create release notes",
    expected_output="GitHub push success and storage report",
    agent=devops
)

print("✅ 9 Tasks defined for v10.0")
```

## Step 6: Create and Run Crew (Company Meeting)

```python
crew = Crew(
    agents=[ceo, cto, kernel_dev, driver_dev, fs_engineer, gui_engineer, userland_engineer, qa_tester, devops],
    tasks=[task1, task2, task3, task4, task5, task6, task7, task8, task9],
    process=Process.sequential,  # sequential for OS build dependencies
    verbose=2
)

# Run the company!
result = crew.kickoff()
print("\n\n=== Zero OS Company Result ===")
print(result)
```
